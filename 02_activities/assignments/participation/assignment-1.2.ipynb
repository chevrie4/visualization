```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(416)
x = np.arange(50)
y1 = np.random.randint(0,100,50)
y2 = np.random.randint(0,100,50)

fig,ax=plt.subplots(figsize=(5,3))
ax.plot(x,y1,label = "person 1")
ax.plot(x,y2,label = "person 2")
ax.legend(loc='lower right',
          frameon = True,
          fontsize = 12,
          ncol =2,
          shadow = True)
ax.text(10,95,"this is cool ---->",
        ha='center',
        color = 'red',
        size = 13)
fig.show()

```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.annotate('This is important!', xy=(10, 95), xytext=(20, 94), arrowprops=dict(arrowstyle="wedge",color="hotpink"))
ax.yaxis.set_major_locator(plt.NullLocator())
ax.xaxis.set_major_formatter(plt.NullFormatter())
fig.show()
```

```python
plt.style.use('fivethirtyeight')
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
```

```python
np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0, 75,50)
x2 = np.array(["Luffy","Zoro","Nami","Usopp","Sanji"])
y2 = np.array([110, 180, 240, 99, 220])
fig, ax = plt.subplots(figsize=(5, 3))
fig,(ax1, ax2) = plt.subplots(ncols=2,nrows=1,figsize=(7, 3))
fig, (ax1, ax2) = plt.subplots(ncols=2, nrows=1, figsize=(7, 3))
ax1.scatter(x1,y1)
ax2.bar(x2,y2)
fig, someaxes = plt.subplot_mosaic([['ax1','ax3'],['ax2','ax3']],figsize=(7, 4))
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
plt.show()
fig.show()
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1','ax3'],['ax2','ax3']],figsize=(7, 4))
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
someaxes["ax1"].set_xlabel('A Big Label',fontsize=10)
someaxes["ax2"].set_xlabel('Another Label',fontsize=10)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious',fontsize=10 )
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1','ax3'],['ax2','ax3']],figsize=(7, 4),layout = "constrained")
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
someaxes["ax1"].set_xlabel('A Big Label'
, fontsize=18)
someaxes["ax2"].set_xlabel('Another Label'
, fontsize=18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious'
, fontsize=18)
```

```python
x = np.array(["Luffy","Zoro","Nami","Usopp","Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])
y2_sd = np.std(y2)
fig, ax = plt.subplots(figsize=(7, 3))
ax.bar(x, y1,color = "indigo")
ax.plot(x, y2,color = "maroon")
ax.errorbar(x, y2, yerr = y2_sd, fmt = "none",ecolor="darkred",elinewidth=4,capsize=6,capthick=4)
```

```python
from wordcloud import WordCloud
df = pd.read_csv("https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/movie_quotes.csv", on_bad_lines = 'skip')
text = " ".join(each for each in df.quote)
wordcloud = WordCloud(background_color= 'white',
                      colormap = 'inferno').generate(text)
fix, ax = plt.subplots(figsize=(7,3))
ax.imshow(wordcloud, # remember 'imshow' from when we added pictures to our matplotlib axes
interpolation='bilinear') # this line helps smooth our image
ax.axis("off")
```
